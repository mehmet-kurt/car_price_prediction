```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error

df = pd.read_csv('car_data.csv')
df.describe()
df.columns
```

```python
car_models = df['name'].to_list()

brands = [car_model.split()[0] for car_model in car_models]

df.insert(0,'brand',brands)

y = df['selling_price']
x = df.drop(columns=['name','selling_price'])

x = pd.get_dummies(x, columns=['brand','fuel','seller_type','transmission','owner'], drop_first=True)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2)

model = RandomForestRegressor()

model.fit(x_train, y_train)

y_pred = model.predict(x_test)

mse = mean_squared_error(y_pred, y_test)
rmse = np.sqrt(mse)

print(rmse)
#print(y_train.to_string())
```

```python
from sklearn.metrics import r2_score
train_r2 = r2_score(y_train, model.predict(x_train))
test_r2 = r2_score(y_test, model.predict(x_test))
print(train_r2)
print(test_r2)
```

```python
import pandas as pd

# Check what the model is actually relying on
if hasattr(model, "feature_importances_"):
    feat_imp = pd.Series(model.feature_importances_, index=x_train.columns)
    print("--- Top 5 Features Driving This 0.96 Score ---")
    print(feat_imp.sort_values(ascending=False).head(5))
```
